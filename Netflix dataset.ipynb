### Task 4: Split the Duration column from the cleaned data, into numeric values

```python
# Extract the number before " min" (movies only; TV shows become NaN)
data_clean['duration_min'] = data_clean['Duration'].str.extract(r'(\d+) min')[0].astype(float)

# Extract the number before " Season" (TV shows only; movies become NaN)
data_clean['duration_season'] = data_clean['Duration'].str.extract(r'(\d+) Season')[0].astype(float)
```

```python
data_clean.head()
```

`Duration` mixes minutes (movies) and seasons (TV shows) as text, so it can't be analysed numerically. I extracted the number with a regex (`\d+`), which handles both "1 Season" and "2 Seasons", and split it into `Duration_Min` (movies only) and `Seasons` (TV shows only). The other category gets NaN, which is what we need

### Task 6: Group ratings into audience categories

```python
rating_map = {
    'TV-Y': 'Kids', 'TV-Y7': 'Kids', 'TV-Y7-FV': 'Kids', 'TV-G': 'Kids', 'G': 'Kids',
    'PG': 'Teens', 'TV-PG': 'Teens', 'PG-13': 'Teens', 'TV-14': 'Teens',
    'R': 'Adults', 'TV-MA': 'Adults', 'NC-17': 'Adults',
    'NR': 'Unrated', 'UR': 'Unrated'
}
# dictionary where every key(rating) is mapped with a value (audience group)
```

```python
data_clean['Rating_Group'] = data_clean['Rating'].map(rating_map)
# map() looks up each Rating in the dictionary and stores the group in a new column (Rating stays unchanged)       
```

```python
data_clean['Rating_Group'].value_counts()
# Count how many titles fall in each group
```

```python
data_clean['Rating_Group'].isnull().sum() # should give the answer as 0
```

The dataset has 14 different ratings (TV and movie systems mixed), which is too huge for comparisons. I grouped them into 4 audience categories (Kids, Teens, Adults, Unrated) in a new `Rating_Group` column. The original `Rating` column is kept unchanged though.

### Q17. Find the most frequent actor pairs.

```python

from itertools import combinations #allows you to generate every possible combination
from collections import Counter # counter helps you to find how often something appears

pair_counts = Counter()  # Stores each actor pair and how often they appear together

for cast in cast_known['Cast'].str.split(', '):
    cast = sorted(set(cast))  # Removes duplicates and sorts actor names
    pair_counts.update(combinations(cast, 2))  # Generates every possible pair

pair_counts.most_common(5)  # Returns the 5 most common pairs
```

### Q19. With respect to the Top 5 countries, calculate: a)average movie duration b)share of each rating group

```python
data_clean.head(10)
```

```python
countries=data_clean[data_clean['Country']!='Unknown']
# Keep known countries
```

```python
countries=countries.assign(Country=countries['Country'].str.split(', ')).explode('Country',ignore_index=True)
#split the multi-country strings, and give every country its own row
# (ignore_index=True resets the row numbers, which avoids a crosstab error from duplicate index table
```

```python
top5=countries['Country'].value_counts().head(5).index
```

```python

# Filter the data to keep only rows belonging to the top 5 countries
top5_data = countries[countries['Country'].isin(top5)]

# Group the data by country, calculate the average movie duration,
# round the averages to 1 decimal place, and arrange countries in top5 order
avg_dur = top5_data.groupby('Country')['duration_min'].mean().round(1).reindex(top5)

# Display the average movie duration for each of the top 5 countries
avg_dur
```

```python
# Calculate the percentage share of each rating group within each country
# normalize='index' converts counts into proportions within each country
# Multiply by 100 to convert proportions into percentages
# Round the percentages to 1 decimal place and maintain the top 5 country order
rating_share = (pd.crosstab(top5_data['Country'], top5_data['Rating_Group'], normalize='index') * 100).round(1).reindex(top5)

# Display the percentage distribution of rating groups for the top 5 countries
rating_share
```

### Q22. Detect outlier movie durations with the IQR method, and then list the outliers

```python
movies=data_clean[data_clean['Category']=='Movie'].copy()
duration=movies['duration_min'].dropna().to_numpy() # dropna drops null values
# Use movies only, as a NumPy array
```

```python
q1,q3=np.percentile(duration,[25,75])
iqr=q3-q1
lower,upper=q1-1.5*iqr,q3+1.5*iqr
print(f"Q1= {q1}, Q3= {q3}, IQR= {iqr}, bounds=({lower},{upper})")
```

```python
outliers_iqr = movies[(movies['duration_min'] < lower) | (movies['duration_min'] > upper)]
print("Total outliers:", len(outliers_iqr))
print("Too Short :", (movies['duration_min']<lower).sum())
print("Too Long :", (movies['duration_min']>upper).sum())
outliers_iqr.sort_values('duration_min', ascending=False)[['Title', 'duration_min']].head(5)
```

### Q23. Compute the Z-scores, by flagging |z| > 3, also print the no of films that are unusally short or unusally long

```python
z_scores=(duration-np.mean(duration))/np.std(duration) # Z score= current value-mean/standard deviation
movies['Z_score']=z_scores.round(2) # inserting z scores column in dataframe
outliers_z=movies[np.abs(movies['Z_score'])>3]
print("Total Flagged : ", len(outliers_z))
print("Too Long : ", (movies['Z_score']>3).sum())
print("Too Short : ", (movies['Z_score']<-3).sum())
outliers_z.sort_values('Z_score', ascending=False)[['Title','duration_min', 'Z_score']].head(5)
```

### Q32. Do TV Shows with more seasons skew toward certain genres or ratings?

```python
# TV shows only, split into 1-season vs 2+ seasons
shows = data_clean[data_clean['Category'] == 'TV Show'].copy()
shows['Season_Group'] = np.where(shows['duration_season'] == 1, '1 Season', '2+ Seasons')
print(shows['Season_Group'].value_counts())

# Rating mix within each group (each column sums to 100%)
print((pd.crosstab(shows['Rating_Group'], shows['Season_Group'], normalize='columns') * 100).round(1))

# Genre mix within each group (explode so every genre gets its own row)
show_genres = shows.assign(Genre=shows['Type'].str.split(', ')).explode('Genre', ignore_index=True)
genre_share = (pd.crosstab(show_genres['Genre'], show_genres['Season_Group'], normalize='columns') * 100).round(1)
genre_share.sort_values('2+ Seasons', ascending=False).head(8)
```

## Key Findings

**Dataset and cleaning**
- The cleaned dataset has ~7,770 titles: ~69% Movies (5,372) and ~31% TV Shows (2,398).
- Director, Cast and Country had thousands of nulls, so they were filled with "Unknown". Fewer than 20 rows (<0.3%) with a missing Rating or Release_Date were dropped.

**Content trends**
- Movies peaked in **2019 (1,498 titles added)** and TV Shows peaked a year later in **2020 (697)**. 2021 is a partial year with respect to this dataset, so its drop is not a real decline.
- There is **no clear shift toward TV Shows**. Their share was 41.8% in 2016, fell to 25.5% in 2018, and recovered to 34.7% in 2020, while movies outnumbered TV Shows every year.
- International Movies, Dramas and Comedies are the top genres overall. International TV Shows, TV Dramas and TV Comedies lead among TV Shows.

**Movie duration**
- The average movie length added per year **rose from ~83 minutes (2015-2016) to ~100-107 minutes (2018 onward)**, so Netflix is adding longer movies, not shorter ones.
- The average movie lasts **99.3 minutes** (median 98, std 28.5). Half of all movies run between 86 and 114 minutes, and only 10% exceed 133 minutes.
- Movie lengths split into **Short (<90 min): 1,649, Medium (90-120): 2,712, Long (>120): 1,011**.
- By rating group, median duration is Kids **69**, Unrated 93, Adults 97, Teens **102** minutes, so kids' content is clearly shorter.
- **336 movies** are IQR outliers (170 too short, 166 too long) and **45** have |z| > 3. These include *Black Mirror: Bandersnatch* (312 min). They are genuine titles, not data errors, so they were kept.

**Countries**
- Indian movies are the **longest (~126 min)** and 70.5% are rated Teens. The US averages 92.4 min and Canada 90.2 min.
- **Canada** has the highest share of kids' content (24.3%), while **France** skews most toward adult ratings (57.9%).
- The **US** produces the most kids' content by volume (375 titles), followed by Canada (100) and the UK (64).

**People**
- Most frequent actors: **Anupam Kher (42 titles)**, Shah Rukh Khan (35), Naseeruddin Shah and Om Puri (30 each), mostly Bollywood.
- Most frequent pair: **John Paul Tremblay and Robb Wells (15 titles)**, followed by the Monty Python members (14 each).

**TV Shows**
- About **two-thirds of TV shows (67%) have only one season**.
- Shows with 2+ seasons skew toward kids' content (25.8% vs 13.5%) and TV Comedies. One-season shows skew toward International TV Shows, Romantic TV Shows and Docuseries.

**Methodology notes**
- `value_counts()` on multi-value columns (Country, Cast, Genre) was misleading until the values were split with `explode()`.
- `np.where` and `pd.cut` gave identical labels once the boundary was handled correctly (`<=120` for Medium). Using `<120` mislabelled the 47 movies of exactly 120 minutes.
